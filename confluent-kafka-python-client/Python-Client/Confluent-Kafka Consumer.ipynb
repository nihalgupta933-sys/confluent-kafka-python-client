```python
from confluent_kafka import Consumer , KafkaError,KafkaException
import json
import time

config = {
  "bootstrap.servers": "pkc-921jm.us-east-2.aws.confluent.cloud:9092",
  "security.protocol": "SASL_SSL",
  "sasl.mechanisms": "PLAIN",
  "sasl.username": "YOUR_CONFLUENT_API_KEY",
  "sasl.password": "YOUR_CONFLUENT_API_SECRET",
  "session.timeout.ms": 45000,
  "client.id": "ccloud-python-client-9eff2c5a-e07c-45cA-a2f0-bda9ea97b2e4",
  'auto.offset.reset': 'earliest',
  'group.id':'customer_grp'
        }
```

```python
consumer=Consumer(config)
```

```python
topic='ecommerce'
consumer.subscribe([topic])

def message_process(message):
  try:
    if message.error():
      if message.error().code==KafkaError._PARTITION_EOF:
        print("End of partition reached {0}/{1} ")
      else:
        raise KafkaException(message.error())
    else:
      key = message.key().decode('utf-8')
      value=json.loads(message.value().decode('utf-8'))

      print(f'Recieved message key {key} value {value}')
  except Exception as e:
    print(f'error consuming/processing error {e}')

```

```python
# poll message continuously
try:
  print('Listening for message , press ctrl + c to exit')
  while True:
    message=consumer.poll(timeout=1.0)
    if(message):
      message_process(message)
except KeyboardInterrupt:
  print("Interupted by user , shutting consumer")
finally:
  consumer.close()

```
